```python
import polars as pl
import matplotlib.pyplot as plt
import numpy as np
from sklearn.linear_model import LinearRegression
```

```python
things = ["izolka", "bateria", "zaciskarka"]
weight = [30, 22.8,  242.1]
voltage = [-476,-352, -3666 ]
```

```python
reg = LinearRegression().fit(np.array(voltage).reshape(-1, 1), weight)
reg.score(np.array(voltage).reshape(-1, 1), weight)

print(f"a: {reg.coef_[0]}, b: {reg.intercept_}")
```

```python
v = np.linspace(-4000, 0, 100)
w = reg.coef_[0] * v + reg.intercept_

fig,ax = plt.subplots()
ax.scatter(voltage, weight, color='red')
ax.plot(v, w, color='blue')
```

```python
#Prediction
predicted_weight = reg.predict(np.array([-1601]).reshape(-1, 1))
print(f"Predicted weight for voltage -1601: {predicted_weight[0]}")
```
